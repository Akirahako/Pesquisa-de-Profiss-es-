```python

saudacao = input("insira seu nome:")
print(f" seja bem vindo {saudacao} ")
cnts = ["usuario", "otario", "sudario"]
slds = {"usuario": 1,
        "otario": 1,
        "sudario":1
        }
try:
  while True :
      print('\n PESQUISA DE VALORES ')
      print('1 : contabilizar dividas')
      print('2 : ver contas ')
      print('3 : depositar ')
      print('4 : sair ')
      esc = int(input('\nescolha uma opçao: '))

      if esc == 4:
          print('obrigado pela preferencia ')
          break

      elif esc == 2:
        print(cnts)
        cnslt = (input('qual conta voce quer consultar '))
        if cnslt in slds:

            print(f'Seu saldo é: R$ {slds[cnslt]}')

        else: print('essa conta nao esta cadastrada')

      elif esc == 3:
        print(cnts)
        escolha_conta = input('qual conta depositar? ')
        if escolha_conta in slds:
          valor = int(input('valor do deposito: '))
          if valor >= 0:
            slds[escolha_conta]= slds[escolha_conta] + valor
            print (f'Seu saldo atualizado é: RS `{slds[escolha_conta]:.2f}')
          else: print('insira um valor valido ')

        else: print('essa conta nao esta cadastrada')



      elif esc ==1:
       print(cnts)
      cnts = (input('qual conta voce quer contabilizar?'))
      if cnts in slds:
              print(f'{slds[cnts]}')
      else: print('essa conta nao esta cadastrada')

      divida = float(input('qual o valor da sua divida? '))
      if divida !=0:
          slds[cnts]= slds[cnts] - divida
          print(f'Seu saldo atualizado é: RS `{slds[cnts]:.2f}')
except Exception:
  print("tomar no seu cu entao")


```
